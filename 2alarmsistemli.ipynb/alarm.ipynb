```python
import numpy as np
import pandas as pd
import time
from sklearn.ensemble import IsolationForest
import os

# -------------------------
# 1. MODEL
# -------------------------
np.random.seed(42)
initial_data = np.random.normal(loc=50, scale=5, size=100).reshape(-1,1)

model = IsolationForest(contamination=0.05)
model.fit(initial_data)

# -------------------------
# 2. SİSTEM BAŞLAT
# -------------------------
print("🚀 Sistem başlatıldı... (Ctrl+C ile durdur)")

try:
    while True:
        # Normal veri üret
        new_value = np.random.normal(50, 5)

        # Arada anomali üret (simülasyon)
        if np.random.rand() < 0.05:
            new_value = np.random.choice([5, 100])

        # Tahmin
        prediction = model.predict([[new_value]])

        # Sonuç yazdır
        status = "ANOMALİ" if prediction[0] == -1 else "Normal"
        print(f"Veri: {round(new_value,2)} → {status}")

        # Alarm
        if prediction[0] == -1:
            print("🚨 DİKKAT! ANOMALİ TESPİT EDİLDİ!")

            # Windows için sesli uyarı
            try:
                import winsound
                winsound.Beep(1000, 500)
            except:
                print("🔊 (Ses çalışmadı ama alarm verildi)")

        

except KeyboardInterrupt:
    print("\n🛑 Sistem durduruldu.")
```
